```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import ast as ast
from IPython.display import HTML
```

```python
data = pd.read_csv("./datasets/ammos.csv")

data["attackPower"] = data["attackPower"].apply(ast.literal_eval)

attack_df = data["attackPower"].apply(
    lambda x: {i["name"]: i["amount"] for i in x}
).apply(pd.Series)


data = pd.concat([data, attack_df], axis=1)

data = data.drop(columns=["Phy20", "Phy40", "Phy100", "attackPower"])
data

```

```python
data[["physical","magic","fire","lightning","holy",	"critical"]].mean().plot(kind="area")
plt.title("Attack Power")
plt.xlabel("Type")
plt.ylabel("Power Level")
plt.show
```

```python
data.fillna({'physical': 0})
```

```python
def path_to_image_html(path):
    return f'<img src = {path} width = 80 height = 40 style = "object-fit: cover;"/>'

HTML(data.to_html(escape=False, formatters={'image': path_to_image_html}))
```
