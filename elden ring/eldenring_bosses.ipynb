```python
import pandas as pd
import numpy as np  
import matplotlib.pyplot as plt
import ast as ast
from IPython.display import HTML
```

```python
bosses = pd.read_csv("./datasets/bosses.csv")

bosses = bosses.drop(columns=["description"])
```

```python
def path_to_image_html(path):
    return f'<img src = {path} width = 80 height = 40 style = "object-fit: cover;"/>'

HTML(bosses.to_html(escape=False, formatters={'image': path_to_image_html}))
```
