```python
#library imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import ast as ast
```

# DATA FRAMING

```python
data=pd.read_csv("./datasets/ammos.csv")
```

```python
#string to list conversion for the attackPower column
data['attackPower']=data['attackPower'].apply(ast.literal_eval)
x=data['attackPower'].apply(lambda y:{i["name"]: i["amount"]for i in y }).apply(pd.Series) 
data=pd.concat([data,x],axis=1) 
```

```python
data=data.drop(columns=['attackPower','Phy20','Phy40','Phy100'])
```

```python
data.head()
```

```python
data.shape
```

```python
print(data['lightning'].max())
```

```python
data=data.dropna(subset=['type','passive'])
```

```python
data.isnull().sum()
```

```python
print(data['lightning'].value_counts())
```

```python
print(data['physical'].value_counts())
```

# all charts

```python
plt.figure(figsize=(10,8))
#lightning attack power distribution
plt.subplot(2,2,1)
plt.hist(data['lightning'],bins=20,color='blue',alpha=0.7)
plt.title("Lightning Attack Power Distribution")
plt.xlabel("Lightning Attack Power")
plt.ylabel("Count")

#fire attack power distribution

plt.subplot(2,2,2)
plt.hist(data['fire'],bins=20,color='red',alpha=0.7)
plt.title("Fire Attack Power Distribution")
plt.xlabel("Fire Attack Power")
plt.ylabel("Count")

#magic attack power distribution

plt.subplot(2,2,3)
plt.hist(data['magic'],bins=20,color='green',alpha=0.7)
plt.title("Magic Attack Power Distribution")
plt.xlabel("Magic Attack Power")
plt.ylabel("Count")

#physical attack power distribution
plt.subplot(2,2,4)
plt.hist(data['critical'],bins=20,color='orange',alpha=0.7)
plt.title("Critical Attack Power Distribution")
plt.xlabel("Critical Attack Power")
plt.ylabel("Count")

plt.tight_layout()
```

```python
data[['lightning', 'fire', 'magic','physical','holy','critical']].mean().plot(kind='barh')
plt.title("Attack Power Distribution")
plt.xlabel("Attack Power Type")
plt.ylabel("Count")
plt.show()
```

```python
data[['lightning', 'fire', 'magic','physical','holy','critical']].mean().plot(kind='box')
plt.title("Attack Power Distribution")
plt.xlabel("Attack Power Type")
plt.ylabel("Count")
plt.show()
```

```python
data[['lightning', 'fire', 'magic','physical','holy','critical']].mean().plot(kind='density')
plt.title("Attack Power Distribution")
plt.xlabel("Attack Power Type")
plt.ylabel("Count")
plt.show()
```

```python
data[['lightning', 'fire', 'magic','physical','holy','critical']].mean().plot(kind='pie')
plt.title("Attack Power Distribution")
plt.xlabel("Attack Power Type")
plt.ylabel("Count")
plt.show()
```

```python
data[['lightning', 'fire', 'magic','physical','holy','critical']].mean().plot(kind='line')
plt.title("Attack Power Distribution")
plt.xlabel("Attack Power Type")
plt.ylabel("Count")
plt.show()
```
