```python
import random
import matplotlib.pyplot as plt

# Generate population
population = []

for i in range(1000):
    population.append(random.randint(1, 100))

sample_means = []

# Draw 100 samples of size 30
for i in range(100):

    sample = []

    for j in range(30):
        sample.append(random.choice(population))

    total = 0
    for value in sample:
        total += value

    mean = total / 30
    sample_means.append(mean)

# Display sample means
print("First 10 Sample Means")
print(sample_means[:10])

# Plot Histogram
plt.hist(sample_means, bins=10, edgecolor='black')

plt.title("Central Limit Theorem")

plt.xlabel("Sample Means")

plt.ylabel("Frequency")

plt.show()
```
