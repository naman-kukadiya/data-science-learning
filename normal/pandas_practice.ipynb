```python
student = pd.DataFrame({
    "Name" : ["Alice", "Bob", "Charlie", "David" , "Eva"], 
    "Age" : [20, 21, 19, 22, 20], 
    "Department": ["CSE", "IT", "CSE", "CE", "IT"], 
    "CGPA": [8.5, 7.8, 9.1, 6.9, 8.8]
})

print(student)
```

```python
print(student.head(3))
print(student.tail(2))
print(student.shape)
print(student.columns)
print(student.dtypes)
print(student.info)
print(student.describe)
print(student["Name"])
print(student[["Name", "CGPA"]])
student["Result"] = student["CGPA"].apply(lambda x : "Passed" if x >=7.0 else "Failed")
print(student)



```

```python
print(student[student["Name"] == "Charlie"])
print(student[0::2])
print(student[student["Department"] == "CSE"])
print(student[student["CGPA"] > 8])
print(student[student["Age"] == 20])
print(student[(student["Department"] == "IT") & (student["CGPA"] > 8)])

print(student.sort_values(by="Age"))
print(student.sort_values(by="CGPA", ascending=False))
print(student.sort_values(by=["Department", "CGPA"], ascending=False))

student["CGPA"] = student["CGPA"] + 0.2
print(student)

student["Department"] = student[student["Name"] == "Bob"] = "CSE"
print(student)

student.rename(columns = {"CGPA": "Grade"})


```
