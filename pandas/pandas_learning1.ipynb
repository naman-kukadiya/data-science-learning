# Problem 1

```python
import pandas as pd 
import matplotlib.pyplot as plt 
import pandas as pd
from scipy.stats import chi2_contingency
plt.figure(figsize=(10,5))
plt.scatter(data['Age'],data['Income'],marker='o')
plt.title("Relationship between age and income")
plt.xlabel("Age")
plt.ylabel("Income")
plt.show()

#correlation between age and income
correlation=data['Age'].corr(data['Income'])
print("Correlation between age and income is:",correlation)
```

# Problem 4

```python
plt.Figure(figsize=(10,5))
data.boxplot(column='Income',by='Education level')
plt.title("Distribution of income by Education level")
plt.xlabel("Education level")
plt.ylabel("Income")
plt.show()
```

# Problem 5

```python
#contingency table 
table=pd.crosstab(data['Education level'],data['Marital Status'])
print(table)

#chi-square test
chi2,p_value,dof,expected=chi2_contingency(table)
print("Chi-square statistic:",chi2)
print("P-value:",p_value)
print("Degrees of freedom:",dof)
```

# Problem 6


```python
data['Age']=pd.to_numeric(data['Age'],errors='coerce')
data=data.dropna(subset=['Age','Education level'])
Education_levels=data['Education level'].unique()
plt.Figure(figsize=(10,5))
for level in Education_levels:
    ages = data[data['Education level']==level]['Age']
    if len(ages)>0:
        plt.hist(ages,bins=10,alpha=0.5,label=level)

plt.title("Distribution of age by Education level")
plt.xlabel("Age")   
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

# Problem 7

```python
table=pd.crosstab(data['Industry'],data['Employee Status'])
table.plot(kind='bar',stacked=True,figsize=(10,5))
plt.title("Employement Status by industry")
plt.xlabel("Industry")
plt.ylabel("Count")
plt.legend(title="Employee Status")
plt.show()
```
